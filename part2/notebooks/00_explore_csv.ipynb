```python
import pandas as pd
import numpy as np
import os

# ============================================
# CHEMINS EXACTS DE VOS FICHIERS
# ============================================

base_path = r'C:\Users\KM-USER\Documents\M1\SII\S2\ResNeur\projectCropClassification\part2\data'

# Lister les fichiers dans le dossier
print("=" * 60)
print("FICHIERS DANS LE DOSSIER:")
print("=" * 60)
for f in os.listdir(base_path):
    if f.endswith('.csv'):
        print(f"  - {f}")

# ============================================
# CHARGER LES FICHIERS CALIFORNIA
# ============================================

# Chercher les fichiers qui contiennent 'california'
ca_files = [f for f in os.listdir(base_path) if 'california' in f.lower() and f.endswith('.csv')]

for file in ca_files:
    filepath = os.path.join(base_path, file)
    df = pd.read_csv(filepath)
    
    print("\n" + "=" * 60)
    print(f"FICHIER: {file}")
    print("=" * 60)
    print(f"Dimensions: {df.shape[0]} lignes, {df.shape[1]} colonnes")
    print(f"\nColonnes:\n{df.columns.tolist()}")
    print(f"\nPremières 3 lignes:")
    print(df.head(3))
    print(f"\nTypes de données:")
    print(df.dtypes)
    
    # Vérifier les valeurs manquantes
    print(f"\nValeurs manquantes: {df.isnull().sum().sum()}")

# ============================================
# ANALYSE DÉTAILLÉE DU FICHIER COVARIATES
# ============================================

# Identifier le fichier qui contient les covariables (celui avec beaucoup de colonnes)
for file in ca_files:
    filepath = os.path.join(base_path, file)
    df = pd.read_csv(filepath)
    if len(df.columns) > 5:  # Le fichier covariables a beaucoup de colonnes
        print("\n" + "=" * 60)
        print("ANALYSE DU FICHIER COVARIATES CALIFORNIA")
        print("=" * 60)
        
        # Plages de valeurs
        print("\n1. Plages de valeurs:")
        if 'climate_temp_mean' in df.columns:
            print(f"   Température: {df['climate_temp_mean'].min():.2f}°C à {df['climate_temp_mean'].max():.2f}°C")
        if 'climate_precip_total' in df.columns:
            print(f"   Précipitations: {df['climate_precip_total'].min():.2f} à {df['climate_precip_total'].max():.2f}")
        if 'soil_ph' in df.columns:
            print(f"   pH du sol: {df['soil_ph'].min():.2f} à {df['soil_ph'].max():.2f}")
        if 'topo_elevation' in df.columns:
            print(f"   Élévation: {df['topo_elevation'].min():.0f}m à {df['topo_elevation'].max():.0f}m")
        
        # Répartition North vs South
        if 'region' in df.columns:
            print("\n2. Répartition North vs South:")
            print(df['region'].value_counts())
        
        # Résumé
        print("\n" + "=" * 60)
        print("CONCLUSION")
        print("=" * 60)
        if df.isnull().sum().sum() == 0:
            print("✅ PAS de valeurs manquantes")
        else:
            print(f"⚠️ {df.isnull().sum().sum()} valeurs manquantes")
        
        print(f"✅ {len(df)} échantillons prêts à être utilisés")
        break
```
