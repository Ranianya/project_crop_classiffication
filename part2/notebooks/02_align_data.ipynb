```python
import numpy as np
import pandas as pd
import os

print("=" * 60)
print("RECRÉATION DES FICHIERS ALIGNÉS AVEC 10,487 POINTS")
print("=" * 60)

# 1. Charger les fichiers sources (Partie 1)
source_path = r'C:\Users\KM-USER\Documents\M1\SII\S2\ResNeur\projectCropClassification\data\processed\california_preprocessed'
X_source = np.load(os.path.join(source_path, 'X.npy'))
y_source = np.load(os.path.join(source_path, 'y.npy'))

print(f"\n📂 Fichiers sources (Partie 1):")
print(f"   X shape: {X_source.shape}")
print(f"   y shape: {y_source.shape}")

# 2. Charger le nouveau CSV (10,487 points)
aligned_path = r'C:\Users\KM-USER\Documents\M1\SII\S2\ResNeur\projectCropClassification\part2\data\aligned'
cov_new = pd.read_csv(os.path.join(aligned_path, 'covariates.csv'))

print(f"\n📂 Nouveau CSV: {cov_new.shape[0]} lignes")

# 3. Vérifier que les nombres correspondent
n_points = cov_new.shape[0]

if X_source.shape[0] != n_points:
    print(f"\n⚠️ Attention: X_source a {X_source.shape[0]} points, CSV a {n_points} points")
    print("   Il faut sous-échantillonner X_source pour correspondre")

# 4. Sous-échantillonner pour correspondre exactement
if X_source.shape[0] > n_points:
    step = X_source.shape[0] // n_points
    X_aligned = X_source[::step][:n_points]
    y_aligned = y_source[::step][:n_points]
    print(f"\n📊 Sous-échantillonnage: 1 pixel sur {step}")
elif X_source.shape[0] < n_points:
    repeats = (n_points // X_source.shape[0]) + 1
    X_aligned = np.tile(X_source, (repeats, 1, 1))[:n_points]
    y_aligned = np.tile(y_source, repeats)[:n_points]
    print(f"\n📊 Répétition pour atteindre {n_points} points")
else:
    X_aligned = X_source
    y_aligned = y_source
    print(f"\n📊 Parfait: nombre exact de points")

print(f"   X_aligned shape: {X_aligned.shape}")
print(f"   y_aligned shape: {y_aligned.shape}")

# 5. Sauvegarder dans aligned/
np.save(os.path.join(aligned_path, 'X_sentinel2.npy'), X_aligned)
np.save(os.path.join(aligned_path, 'y_labels.npy'), y_aligned)

print(f"\n✅ Fichiers sauvegardés dans {aligned_path}")
print(f"   - X_sentinel2.npy: {X_aligned.shape}")
print(f"   - y_labels.npy: {y_aligned.shape}")

# 6. Vérification finale
print("\n" + "=" * 60)
print("VÉRIFICATION FINALE")
print("=" * 60)
print(f"X_sentinel2 shape: {np.load(os.path.join(aligned_path, 'X_sentinel2.npy')).shape}")
print(f"y_labels shape: {np.load(os.path.join(aligned_path, 'y_labels.npy')).shape}")
print(f"covariates shape: {pd.read_csv(os.path.join(aligned_path, 'covariates.csv')).shape}")

print("\n✅ TOUT EST PRÊT ! Vous pouvez lancer l'ablation study.")
```
